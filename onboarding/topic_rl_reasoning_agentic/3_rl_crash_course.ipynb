## 1. States, actions, and transitions
An agent selects an action; the environment returns a next state, reward, and termination flag. Our corridor has states 0–3. Start at 0; reaching 3 ends the episode. Actions are `left` and `right`. Each move costs −1, except entering the goal earns +5.

**Predict:** What happens when you move left at state 0?

```python
def step(state, action):
    if state == 3:
        raise ValueError("Reset after termination")
    if action not in ("left", "right"):
        raise ValueError("Unknown action")
    next_state = max(0, min(3, state + (1 if action == "right" else -1)))
    done = next_state == 3
    return next_state, 5 if done else -1, done

print(step(0, "left"))
print(step(2, "right"))
```

## 4. Exploration and evaluating an agent
An epsilon-greedy policy explores with probability epsilon; otherwise it chooses a highest-value action. Training includes exploration, while a greedy evaluation can measure the learned policy. Use a step limit so a poor policy cannot loop forever.

**Predict:** Why can a single successful episode hide a poor policy?

```python
import random
rng = random.Random(0)
actions = ("left", "right")
Q = {s: {a: 0.0 for a in actions} for s in range(4)}

def choose_action(state, epsilon):
    if rng.random() < epsilon:
        return rng.choice(actions)
    return max(actions, key=lambda a: Q[state][a])
```
